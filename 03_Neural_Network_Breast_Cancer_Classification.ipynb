```python
import numpy as np
import pandas as pd
import torch
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import LabelEncoder
```

```python
df = pd.read_csv('https://raw.githubusercontent.com/gscdit/Breast-Cancer-detection/refs/heads/master/data.csv')
```

```python
df.drop(columns=['id','Unnamed: 32'],inplace=True)
```

### Train Test Split

```python
X_train, X_test, y_train, y_test = train_test_split(df.iloc[:,1:],df.iloc[:,0],test_size=0.2)
```

### Scaling

```python
scaler = StandardScaler()

# 1. Train data par fit aur transform dono karein
X_train = scaler.fit_transform(X_train)

# 2. Test data par SIRF transform karein (taaki train wala mean/std hi use ho)
X_test = scaler.transform(X_test)
```

```python
y_train
```

### Label Encoding

```python
encoder = LabelEncoder()
y_train = encoder.fit_transform(y_train)
y_test = encoder.transform(y_test)
```

```python
y_train
```

### Numpy Arrays to PyTorch Tensors

```python
X_train_tensors = torch.from_numpy(X_train)
X_test_tensors = torch.from_numpy(X_test)
y_train_tensors = torch.from_numpy(y_train)
y_test_tensors = torch.from_numpy(y_test)
```

```python
X_train_tensors.shape
```

```python
y_train_tensors.shape
```

### Defining the Model

```python
import torch.nn as nn
import torch.optim as optim


class MySimpleNN(nn.Module):
  def __init__(self, X):
        super(MySimpleNN, self).__init__()  # nn.Module ko initialize karna zaroori hai

        # Weights aur Bias ko nn.Parameter banana padega taaki model.parameters() kaam kare
        self.weights = nn.Parameter(torch.rand(X.shape[1], 1, dtype=torch.float64))
        self.bias = nn.Parameter(torch.zeros(1, dtype=torch.float64))


  def forward(self,X):
    z = torch.matmul(X, self.weights) + self.bias
    y_pred = torch.sigmoid(z)
    return y_pred


  def loss_function(self,y_pred,y):
    # Clamp prediction to avoid log(0)
    epsilon = 1e-7
    y_pred = torch.clamp(y_pred, epsilon, 1 - epsilon)
    # caluculate loss
    loss = - (y * torch.log(y_pred) + (1 - y) * torch.log(1 - y_pred)).mean()
    return loss
```

### Important Parameters

```python
learning_rate = 0.05
epochs = 1000
```

### Training Pipeline

```python
# Create model
model = MySimpleNN(X_train_tensors)

# Ab optimizer define karein (Ab model.parameters() perfectly kaam karega)
optimizer = optim.SGD(model.parameters(), lr=0.01)

# define in loop = epochs
for epochs in range(epochs):
# forward pass
  y_pred = model.forward(X_train_tensors)

# loss calculate
  loss = model.loss_function(y_pred, y_train_tensors)

# bachward pass
  loss.backward()

# parameters update
  with torch.no_grad():
    model.weights -= learning_rate * model.weights.grad
    model.bias -= learning_rate * model.bias.grad

# Zero gradients
  model.weights.grad.zero_()
  model.bias.grad.zero_()

# print loss in each epochs
  print(f'Epochs : {epochs + 1}, loss : {loss.item()}')
```

### Model Evaluation

```python
with torch.no_grad():
    y_pred = model.forward(X_test_tensors)

    # Threshold ko 0.5 karein (Standard Binary Classification rule)
    y_pred = (y_pred >= 0.5).float()

    accuracy = (y_pred == y_test_tensors).float().mean()
    print(f'Model Accuracy : {accuracy.item() * 100:.2f}%')
```

Improve model Accuracy

```python
import torch
import torch.nn as nn
import torch.optim as optim

class MyUpgradedNN(nn.Module):
    def __init__(self, input_dim):
        super(MyUpgradedNN, self).__init__()
        # Input layer se Hidden layer (16 neurons)
        self.fc1 = nn.Linear(input_dim, 16)
        # Hidden layer se Output layer (1 neuron)
        self.fc2 = nn.Linear(16, 1)

    def forward(self, X):
        # Hidden layer ke baad non-linear ReLU activation
        x = torch.relu(self.fc1(X))
        # Final output ke liye Sigmoid function
        y_pred = torch.sigmoid(self.fc2(x))
        return y_pred
```

```python
# 1. Input dimensions pata karein (Aapke features ki ginti)
input_dim = X_train_tensors.shape[1]

# 2. Model, Loss aur Optimizer initialize karein
model = MyUpgradedNN(input_dim).double()
criterion = nn.BCELoss()  # PyTorch ka standard Binary Cross Entropy Loss
optimizer = optim.Adam(model.parameters(), lr=0.005)  # Adam optimizer standard SGD se fast hai

# 3. Epochs ko badha kar 1000 ya 1500 karein
num_epochs = 1000

for epoch in range(num_epochs):
    # Forward Pass
    y_pred = model(X_train_tensors)

    # Loss Calculation
    loss = criterion(y_pred, y_train_tensors.unsqueeze(1).double())

    # Backward Pass aur Weights Update
    optimizer.zero_grad()  # Pichle gradients saaf karein
    loss.backward()        # Backpropagation
    optimizer.step()       # Weights update karna

    # Har 100 epochs baad loss print karein
    if (epoch + 1) % 100 == 0:
        print(f'Epoch: {epoch + 1}/{num_epochs}, Loss: {loss.item():.4f}')
```

```python
with torch.no_grad():
    # 1. Test data par prediction lein
    y_pred_test = model(X_test_tensors)

    # 2. Threshold 0.5 set karein
    y_pred_class = (y_pred_test >= 0.5).float()

    # 3. Target ko unsqueeze karke exact same shape (2D) me compare karein
    accuracy = (y_pred_class == y_test_tensors.unsqueeze(1)).float().mean()

    print(f'Upgraded Model Accuracy : {accuracy.item() * 100:.2f}%')
```
